```python
import numpy as np
import pandas as pd
from sklearn.linear_model import Ridge
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# 1. Load Dataset
df = pd.read_csv('household_data_60min_singleindex.csv')

# 2. Select Required Columns for Residential 3
cols = [
    'utc_timestamp',
    'DE_KN_residential3_grid_import',
    'DE_KN_residential3_grid_export',
    'DE_KN_residential3_pv',
]
df_subset = df[cols].copy()
df_subset['utc_timestamp'] = pd.to_datetime(df_subset['utc_timestamp'])
df_subset = df_subset.sort_values('utc_timestamp').reset_index(drop=True)

# 3. Clean Missing Values (Fill NaN grid/PV metrics with 0)
df_subset[
    [
        'DE_KN_residential3_grid_import',
        'DE_KN_residential3_grid_export',
        'DE_KN_residential3_pv',
    ]
] = df_subset[
    [
        'DE_KN_residential3_grid_import',
        'DE_KN_residential3_grid_export',
        'DE_KN_residential3_pv',
    ]
].fillna(
    0
)

# 4. Calculate Total House Load (Target)
df_subset['total_load'] = (
    df_subset['DE_KN_residential3_grid_import']
    + df_subset['DE_KN_residential3_pv']
    - df_subset['DE_KN_residential3_grid_export']
)

# 5. Feature Engineering (Lag features to predict tomorrow's load from past data)
df_subset['hour'] = df_subset['utc_timestamp'].dt.hour
df_subset['dayofweek'] = df_subset['utc_timestamp'].dt.dayofweek
df_subset['month'] = df_subset['utc_timestamp'].dt.month

df_subset['load_lag_24'] = df_subset['total_load'].shift(
    24
)  # Same hour yesterday
df_subset['load_lag_48'] = df_subset['total_load'].shift(
    48
)  # 2 days ago
df_subset['load_lag_168'] = df_subset['total_load'].shift(
    168
)  # Same hour last week

# Drop rows with NaN from lagging
df_clean = df_subset.dropna().reset_index(drop=True)

features = ['hour', 'dayofweek', 'month', 'load_lag_24', 'load_lag_48', 'load_lag_168']
target = 'total_load'

X = df_clean[features]
y = df_clean[target]

# 6. Chronological Train/Test Split (80% Train, 20% Test)
train_size = int(len(df_clean) * 0.8)
X_train, X_test = X.iloc[:train_size], X.iloc[train_size:]
y_train, y_test = y.iloc[:train_size], y.iloc[train_size:]

# 7. Train Best Model (Ridge Regression)
model = Ridge(alpha=1.0)
model.fit(X_train, y_train)

# 8. Predict & Evaluate
y_pred = model.predict(X_test)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
mae = mean_absolute_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print('--- Load Forecasting Results ---')
print(f'Test RMSE: {rmse:.2f} W')
print(f'Test MAE:  {mae:.2f} W')
print(f'Test R2 Score: {r2:.4f}')
print(f'Model Accuracy: {r2 * 100:.2f}%')
```
